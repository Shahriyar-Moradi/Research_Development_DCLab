# 06 — Interactions and bins

**Goal:** combine pricing + time + geo features, then add interactions and quantile bins.

## Why this experiment?
Some effects are joint (e.g. long distance **and** low fare).
Quantile bins also help trees capture non-linear cutoffs.

## Approach
1. Build the full engineered set (pricing + time + geo).
2. Add interactions: distance × first fare, category × hour.
3. Fit quantile bins on **train only**, apply to test.
4. Train LightGBM.

## What changed?
- Features: all FE from 03–05 plus interactions and bins
- Still one model family (LightGBM), no tuning yet

## Features used in this notebook
- Full pricing + time + geo features
- **Extra:** distance×fare, category×hour, quantile bins of distance & first fare
- **How selected:** combine winning FE families, then add joint effects and soft cutoffs
- **Why:** best pure feature-engineering experiment before model tuning


### Setup
Shared data load and fixed split.


```python
import os
from pathlib import Path
import sys

import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

for candidate in (Path.cwd(), *Path.cwd().parents):
    if (candidate / "shared" / "protocol.py").exists():
        EXPERIMENT_ROOT = candidate
        break
    if (candidate / "hyperack_exp" / "shared" / "protocol.py").exists():
        EXPERIMENT_ROOT = candidate / "hyperack_exp"
        break
else:
    raise RuntimeError("Run this notebook from the HyperAck project directory.")
os.chdir(EXPERIMENT_ROOT)
sys.path.insert(0, str(EXPERIMENT_ROOT))

from shared.protocol import (
    add_geo_features,
    add_pricing_features,
    add_time_features,
    base_features,
    evaluate,
    load_clean_df,
    make_xy,
    save_result,
    split_frame,
    actual_vs_predicted_report,
)

RANDOM_STATE = 42
train_df, test_df = split_frame(load_clean_df())
```

### Feature engineering
Full FE + interactions + train-fitted quantile bins.


```python
from sklearn.preprocessing import KBinsDiscretizer
X_train, y_train = make_xy(train_df, pricing=True, time_features=True, geo=True)
X_test, y_test = make_xy(test_df, pricing=True, time_features=True, geo=True)
X_train["distance_x_first_fare"] = X_train["total_distance"] * X_train["first_customer_fare"]
X_test["distance_x_first_fare"] = X_test["total_distance"] * X_test["first_customer_fare"]
X_train["category_x_hour"] = X_train["deliverey_category_id"] * X_train["hour"]
X_test["category_x_hour"] = X_test["deliverey_category_id"] * X_test["hour"]
bin_cols = ["total_distance", "first_customer_fare"]
binning = Pipeline([("imputer", SimpleImputer(strategy="median")), ("bin", KBinsDiscretizer(n_bins=8, encode="ordinal", strategy="quantile"))])
bins_train = binning.fit_transform(X_train[bin_cols])
bins_test = binning.transform(X_test[bin_cols])
for idx, name in enumerate(bin_cols):
    X_train[f"{name}_qbin"] = bins_train[:, idx]
    X_test[f"{name}_qbin"] = bins_test[:, idx]
```

### Features selected and why

This is the **richest feature set** before tuning:

1. Base raw columns  
2. Pricing transforms  
3. Time / cyclical transforms  
4. Geo distance & bearing  
5. Interactions: `distance_x_first_fare`, `category_x_hour`  
6. Quantile bins fit on **train only**

| Extra idea | Why |
|------------|-----|
| Interactions | Two things together can matter (long + cheap) |
| Quantile bins | Soft thresholds like “very long” / “very cheap” without hard-coding cuts |

Next cell lists every selected column.

### Model
LightGBM on the richer feature matrix.


```python
from lightgbm import LGBMClassifier
model = Pipeline([("imputer", SimpleImputer(strategy="median")), ("model", LGBMClassifier(n_estimators=800, learning_rate=0.035, num_leaves=31, reg_lambda=1.0, random_state=RANDOM_STATE, n_jobs=-1, verbosity=-1))])
```

### Evaluate
Held-out metrics.


```python
metrics = evaluate(model, X_train, y_train, X_test, y_test)
metrics
```
